## Cleaning income columns

Define a list of those 14 column names

Replace 99999999999 with NaN across all of them

Replace any blank strings with NaN

Convert to numeric type

Sum them into a new column INGRESO_TOTAL, using skipna=True

Where all 14 columns are NaN, set INGRESO_TOTAL to NaN

```python
# All E... columns are set as income_cols

income_cols = [
    'E01AIMDE',
    'E01BIMDE',
    'E01CIMDE',
    'E01DDE',
    'E01EDE',
    'E01FDE',
    'E01GDE',
    'E01HDE',
    'E01IDE',
    'E01JDE',
    'E01KDE',
    'E01LDE',
    'E01MDE',
    'E01KJDE'
]
```

```python
# Replacing local thousands separators by standardized ones.

for col in income_cols:
    df[col] = df[col].str.replace(',', '.')
```

```python
# Changing income columns to numeric types.

for col in income_cols:
    df[col] = pd.to_numeric(df[col], errors = 'coerce')

df.head(20)
```

```python
# Verifying if changes applied correctly.

df.dtypes
```

```python
# Verifying that there are no longer sentinel values.

print((df[income_cols] == 99999999999).sum())
print("\n")
print((df[income_cols] == "99999999999").sum())
```

```python
# Replacing 0s with NaN in income columns

df[income_cols] = df[income_cols].replace(0, np.nan)
```

```python
# Creating a total_income column to sum all types of income

df['TOTAL_INCOME'] = df[income_cols].sum(axis = 1)
```

## Filtering for working age population

```python
# The standard INE uses for the EPHC is 15 years old, that's also the international ILO standard for labor force surveys.

df_workingage = df[df['P02'] >= 15].copy()
df_workingage.shape
```

```python
df_workingage.head()
```

```python
# df_workingage[df_workingage['PEAA'] == '1'].groupby('DPTO_NAMES')['TOTAL_INCOME'].sum()

# Validating logic for calculating employment rate, testing for Asuncion first.

asuncion_peaa_1 = df_workingage.loc[(df_workingage['DPTO'] == 0) & (df_workingage['PEAA'] == '1'), 'Factor'].sum()
asuncion_peaa_1_2 = df_workingage.loc[(df_workingage['DPTO'] == 0) & ((df_workingage['PEAA'] == '1') | (df_workingage['PEAA'] == '2')), 'Factor'].sum()

print(asuncion_peaa_1 / asuncion_peaa_1_2 * 100.0)
```

```python
# With logic validated, expanding the calculation for the other departments.

emp_rate = df_workingage[df_workingage['PEAA'] == '1'].groupby('DPTO_NAMES')['Factor'].sum() / df_workingage[(df_workingage['PEAA'] == '1') | (df_workingage['PEAA'] == '2')].groupby('DPTO_NAMES')['Factor'].sum() * 100.0
emp_rate
```

```python
# Calculating the weighted average income by department: starting with creating an intermediate column that multiplies total_income * factor

df_workingage['TOTAL_INCOME_BY_FACTOR'] = df_workingage['TOTAL_INCOME'] * df_workingage['Factor']
df_workingage.head()
```

```python
avg_income = df_workingage[df_workingage['PEAA'] == '1'].groupby('DPTO_NAMES')['TOTAL_INCOME_BY_FACTOR'].sum() / df_workingage[df_workingage['PEAA'] == '1'].groupby('DPTO_NAMES')['Factor'].sum()
avg_income
```

```python
URBAN_POVERTY_THRESHOLD = 897168
RURAL_POVERTY_THRESHOLD = 654657
```

For poverty headcount the logic is:

1. For each person (row), check if TOTAL_INCOME is below the poverty line threshold
2. If yes, that person is poor
3. To get the weighted count of poor people, sum their Factor values
4. Divide by total Factor in that department to get the poverty rate

```python
is_poor_urban_mask = (df_workingage['AREA'] == 1) & (df_workingage['TOTAL_INCOME'] < URBAN_POVERTY_THRESHOLD)
print(is_poor_urban_mask.value_counts())
is_poor_rural_mask = (df_workingage['AREA'] == 6) & (df_workingage['TOTAL_INCOME'] < RURAL_POVERTY_THRESHOLD)
print(is_poor_rural_mask.value_counts())
is_poor_mask = is_poor_rural_mask | is_poor_urban_mask
print(is_poor_mask.value_counts())
```

```python
poverty_headcount = (df_workingage.loc[is_poor_mask].groupby('DPTO_NAMES')['Factor'].sum() / df_workingage.groupby('DPTO_NAMES')['Factor'].sum() * 100.0).sort_values(ascending = False)
poverty_headcount
```

```python
emp_inc_pov = pd.concat([emp_rate, avg_income, poverty_headcount], axis = 1)
emp_inc_pov.columns = ['Employment Rate', 'Average Income', 'Poverty Headcount']
emp_inc_pov.sort_values(by = ['Poverty Headcount'], ascending = False)
```
